```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv('healthcare-dataset-stroke-data.csv')
df.info()
df.head(10)
```

```python
df.describe(include = "all")
```

```python
sns.countplot(x = "stroke", data = df)
plt.title("distribusi kelas stroke")
plt.show()
```

```python
df.dtypes
```

```python
cat_col = ["gender", "ever_married", "work_type", "Residence_type", "smoking_status"]

for col in cat_col:
    df[col] = df[col].astype('category')

df.dtypes
```

```python
df.isnull().sum()
```

```python
df['bmi'] = df['bmi'].fillna(df['bmi'].median())

df.head()
```

```python
df.isnull().sum()
```

```python
sns.boxplot(x="bmi", data = df)
plt.title("distribusi kelas bmi")
plt.show()
```

```python
Q1 = df['age'].quantile(0.25)
Q3 = df['age'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df["age"] < lower_bound) | (df["age"] > upper_bound)]
print("jumlah outlier umur :", len(outliers))
```

### Correlation variable

```python
num_df = df.select_dtypes(include = ['int64', 'float64'])

plt.figure(figsize = (10,6))
sns.heatmap(num_df.corr(), annot = True, cmap = "coolwarm")
plt.show()
```

```python
sns.boxplot(x="stroke", y ="age", data = df)
plt.title("hubungan umur dengan stroke")
plt.show()
```
